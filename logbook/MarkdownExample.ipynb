## Project overview

This project extends earlier DT4PED research from district-scale analysis toward a broader city-scale and cross-country tool for decision support. It supports building-stock analysis, retrofit prioritisation and urban planning by combining multiple datasets with transparent assumptions.

## Data sources and rationale

The tool combines several data sources because each layer serves a different analytical purpose.

- EUBUCCO: Swedish building geometry and attributes.
- OpenStreetMap: UK footprint geometry and address-based matching.
- Swedish energideklaration register: EPC and building energy performance evidence.
- UK EPC API: official performance certificates where authorisation is available.
- TABULA / EPISCOPE: archetype assumptions where measured data are absent.
- DTCC LiDAR: vegetation, roof form and terrain context.
- Wikells and Boverket: cost and carbon factors.
- EPW weather files: climate forcing for EnergyPlus simulation.
- Boplats / Booli / Trafikverket / Västtrafik: market and urban context signals.

## Transparency and limitations

This project intentionally separates raw data, processed outputs and assumptions. The logbook documents known gaps and uncertainties so that results are not presented as more precise than they are. This includes archetype fallback logic, country-specific matching approaches, missing metadata and synthetic placeholders where no real data exist.
