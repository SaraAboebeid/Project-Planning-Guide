# Project Team & Credits

This notebook summarises the project lineage, who is involved, what was adapted from earlier work, and why each dataset and method is used in the project. The purpose is to make the logbook traceable, transparent and aligned with the evidence-based structure used in the DT4PED-style documentation.

---

## 1. Project context

This project sits in the DT4PED lineage: a research and software effort focused on digital twins for positive-energy districts and decision support for building and urban transformation. The present tool extends that logic from district-level studies to broader city-scale and cross-country analysis.

The design choices recorded in the project are therefore shaped by three practical constraints:

- city-scale analysis requires a highly automated and computationally tractable workflow;
- broader geographic coverage requires open and interoperable data sources;
- decision support demands transparent assumptions and clear documentation of limitations.

## 3. Project lineage and technical framing

The project is not only a front-end interface. It is an end-to-end decision-support environment that combines:

- building stock data and open registers;
- digital twin and geospatial visualisation;
- simulation and performance assessment;
- cost and carbon reasoning;
- optimisation and prioritisation workflows.

This makes the project a research-to-practice tool, where the main value is not the interface alone but the traceability of the entire decision path from raw data to recommendation.

## 4. Data sources and rationale

The project relies on a heterogeneous but explicit evidence base. The logic behind each source is practical and traceable.

| Source / component | Why it is used | Why it matters here |
|---|---|---|
| EUBUCCO | Building geometry and attributes for Sweden | Primary footprint source for Swedish building stock analysis |
| OpenStreetMap | Footprints and address metadata for the UK | Provides the geometry and address keys needed for UK matching logic |
| Swedish energideklaration register | Energy performance certificates for Swedish buildings | Enables coverage checks, benchmarking and matching with property-level energy data |
| UK EPC open-data service | Official UK energy certificate data | Supports direct performance evidence when an access token is available |
| TABULA / EPISCOPE | Building archetype assumptions | Fills data gaps where measured building attributes are absent |
| DTCC LiDAR | Terrain, vegetation and roof-form data | Improves environmental analysis and realism of the digital twin |
| EPW weather files | Climate forcing for simulation | Required to run EnergyPlus-based thermal and environmental analysis |
| Wikells + Boverket | Retrofit cost and carbon references | Supports decision analysis on cost and embodied impact |
| Boplats / Booli / Trafikverket / Västtrafik | Market and urban context data | Adds the broader urban system view needed for practical planning decisions |

The project therefore follows a clear provenance principle: data that enters the workflow must be documented, traceable and auditable.

## 6. Transparency and limitations

The project is intentionally honest about uncertainty. Several parts of the workflow rely on assumptions rather than exact measurement, and the logbook records these explicitly. This includes:

- missing building attributes;
- archetype fallback logic;
- country-specific matching methods;
- incomplete or synthetic placeholders where real data are unavailable;
- the need to distinguish provisional outputs from final results.

This is a key reason the logbook format is not simply a description page — it is a traceability document for method quality and source reliability.
